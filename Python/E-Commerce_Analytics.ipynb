```python
# Extract a business KPI using Python
query = """
SELECT
    COUNT(*) AS total_orders,
    ROUND(SUM(CASE
        WHEN order_status = 'Delivered'
        THEN order_value ELSE 0
    END), 2) AS delivered_revenue,
    ROUND(AVG(CASE
        WHEN order_status = 'Delivered'
        THEN order_value
    END), 2) AS average_order_value
FROM orders;
"""

kpi_df = pd.read_sql(query, engine)

kpi_df
```

```python
# Load delivered orders into Pandas
query = """
SELECT
    order_id,
    customer_id,
    order_date,
    order_status,
    order_value
FROM orders
WHERE order_status = 'Delivered';
"""

orders_df = pd.read_sql(query, engine)

print("Total records:", len(orders_df))
orders_df.head()
```

```python
# Calculate business KPIs using Pandas
total_orders = len(orders_df)

total_revenue = orders_df["order_value"].sum()

average_order_value = orders_df["order_value"].mean()

unique_customers = orders_df["customer_id"].nunique()

print("Total Delivered Orders:", total_orders)
print("Total Delivered Revenue:", round(total_revenue, 2))
print("Average Order Value:", round(average_order_value, 2))
print("Unique Customers:", unique_customers)
```

```python
# Monthly revenue analysis
# Extract month from order_date
orders_df["month"] = orders_df["order_date"].dt.to_period("M")

# Group by month
monthly_analysis = orders_df.groupby("month").agg(
    total_orders=("order_id", "count"),
    total_revenue=("order_value", "sum"),
    average_order_value=("order_value", "mean")
).reset_index()

# Round monetary values
monthly_analysis["total_revenue"] = monthly_analysis["total_revenue"].round(2)
monthly_analysis["average_order_value"] = monthly_analysis["average_order_value"].round(2)

monthly_analysis.head(10)
```

```python
# Calculate MoM revenue growth
# Calculate month-over-month revenue growth
monthly_analysis["mom_growth_pct"] = (
    monthly_analysis["total_revenue"].pct_change() * 100
)

monthly_analysis["mom_growth_pct"] = (
    monthly_analysis["mom_growth_pct"].round(2)
)

monthly_analysis
```

```python
# Category-wise revenue and profitability analysis
query = """
SELECT
    p.category,
    SUM(oi.quantity) AS total_units,
    ROUND(SUM(oi.quantity * oi.unit_price), 2) AS revenue,
    ROUND(SUM(oi.quantity * oi.unit_cost), 2) AS total_cost,
    ROUND(
        SUM(oi.quantity * oi.unit_price) -
        SUM(oi.quantity * oi.unit_cost), 2
    ) AS gross_profit
FROM orders o
JOIN order_items oi
    ON o.order_id = oi.order_id
JOIN products p
    ON oi.product_id = p.product_id
WHERE o.order_status = 'Delivered'
GROUP BY p.category
ORDER BY revenue DESC;
"""

category_df = pd.read_sql(query, engine)

category_df
```

```python
# Calculate gross profit margin for each category
category_df["gross_profit_margin_pct"] = (
    category_df["gross_profit"] / category_df["revenue"] * 100
).round(2)

category_df
```

```python
# Customer-level performance analysis
query = """
SELECT
    customer_id,
    COUNT(order_id) AS total_orders,
    SUM(order_value) AS total_revenue,
    AVG(order_value) AS average_order_value,
    MIN(order_date) AS first_order_date,
    MAX(order_date) AS last_order_date
FROM orders
WHERE order_status = 'Delivered'
GROUP BY customer_id;
"""

customer_df = pd.read_sql(query, engine)

customer_df.head(10)
```

```python
# Analyse customer performance
print("Total customers:", len(customer_df))

print("Average customer revenue:",
      round(customer_df["total_revenue"].mean(), 2))

print("Average orders per customer:",
      round(customer_df["total_orders"].mean(), 2))

print("Highest customer revenue:",
      round(customer_df["total_revenue"].max(), 2))
```

```python
# Customer performance metrics
print("Total customers:", len(customer_df))

print("Average customer revenue:",
      round(customer_df["total_revenue"].mean(), 2))

print("Average orders per customer:",
      round(customer_df["total_orders"].mean(), 2))

print("Highest customer revenue:",
      round(customer_df["total_revenue"].max(), 2))
```

```python
# Identify high-value customers
top_customers = customer_df.sort_values(
    by="total_revenue",
    ascending=False
).head(10)

top_customers
```

```python
# Segment customers by purchase frequency
import numpy as np

conditions = [
    customer_df["total_orders"] == 1,
    customer_df["total_orders"].between(2, 3),
    customer_df["total_orders"].between(4, 5),
    customer_df["total_orders"] > 5
]

segments = [
    "One-time",
    "Occasional",
    "Frequent",
    "Loyal"
]

customer_df["customer_segment"] = np.select(
    conditions,
    segments,
    default="Other"
)

customer_df.head(10)
```

```python
# Calculate segment distribution
segment_analysis = customer_df.groupby(
    "customer_segment"
).agg(
    customers=("customer_id", "count"),
    average_revenue=("total_revenue", "mean"),
    total_revenue=("total_revenue", "sum")
).reset_index()

segment_analysis["customer_percentage"] = (
    segment_analysis["customers"] /
    segment_analysis["customers"].sum() * 100
).round(2)

segment_analysis
```

```python
# Customer Retention and Repeat Purchase Rate.
```

```python
# Identify repeat customers
repeat_customers = customer_df[
    customer_df["total_orders"] > 1
]

one_time_customers = customer_df[
    customer_df["total_orders"] == 1
]

total_customers = len(customer_df)

repeat_rate = (
    len(repeat_customers) / total_customers
) * 100

print("Total customers:", total_customers)
print("Repeat customers:", len(repeat_customers))
print("One-time customers:", len(one_time_customers))
print("Repeat purchase rate:", round(repeat_rate, 2), "%")
```

```python
# Analyse revenue contribution by customer segment
segment_analysis["revenue_contribution_pct"] = (
    segment_analysis["total_revenue"] /
    segment_analysis["total_revenue"].sum() * 100
).round(2)

segment_analysis
```

```python
# Marketing Channel Performance and CAC Analysis.
```

```python
# Extract marketing spend by channel
```

```python
query = """
SELECT
    channel,
    SUM(spend) AS total_spend,
    COUNT(*) AS campaign_records
FROM marketing_spend
GROUP BY channel;
"""

marketing_df = pd.read_sql(query, engine)

marketing_df
```

```python
# Extract customers acquired by channel
query = """
SELECT
    acquisition_channel,
    COUNT(*) AS acquired_customers
FROM customers
GROUP BY acquisition_channel;
"""

acquisition_df = pd.read_sql(query, engine)

acquisition_df
```

```python
# Calculate preliminary CAC
cac_df = marketing_df.merge(
    acquisition_df,
    left_on="channel",
    right_on="acquisition_channel"
)

cac_df["CAC"] = (
    cac_df["total_spend"] /
    cac_df["acquired_customers"]
).round(2)

cac_df = cac_df.sort_values(
    by="CAC",
    ascending=True
)

cac_df[[
    "channel",
    "total_spend",
    "acquired_customers",
    "CAC"
]]
```

```python
# Merge marketing spend and customer acquisition
cac_df = marketing_df.merge(
    acquisition_df,
    left_on="channel",
    right_on="acquisition_channel",
    how="inner"
)

cac_df
```

```python
# Calculate channel-wise customer revenue and profitability
query = """
SELECT
    c.acquisition_channel,
    COUNT(DISTINCT c.customer_id) AS customers,
    SUM(o.order_value) AS total_revenue,
    COUNT(o.order_id) AS total_orders
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
WHERE o.order_status = 'Delivered'
GROUP BY c.acquisition_channel
ORDER BY total_revenue DESC;
"""

channel_performance_df = pd.read_sql(query, engine)

channel_performance_df
```

```python
# Calculate revenue per delivered-order customer:
channel_performance_df["revenue_per_customer"] = (
    channel_performance_df["total_revenue"] /
    channel_performance_df["customers"]
).round(2)

channel_performance_df
```

```python
# Merge marketing and acquisition data
cac_df = marketing_df.merge(
    acquisition_df,
    left_on="channel",
    right_on="acquisition_channel",
    how="inner"
)

cac_df
```

```python
# Calculate Customer Acquisition Cost (CAC)
cac_df["CAC"] = (
    cac_df["total_spend"] /
    cac_df["acquired_customers"]
).round(2)

cac_df = cac_df.sort_values(
    by="CAC",
    ascending=True
)

cac_df[[
    "channel",
    "total_spend",
    "acquired_customers",
    "CAC"
]]
```

```python
# Revenue-to-CAC ratio
```

```python
# Calculate revenue per acquired customer and Revenue/CAC
cac_df = cac_df.merge(
    channel_performance_df[[
        "acquisition_channel",
        "total_revenue"
    ]],
    left_on="channel",
    right_on="acquisition_channel",
    how="left"
)

cac_df["revenue_per_customer"] = (
    cac_df["total_revenue"] /
    cac_df["acquired_customers"]
)

cac_df["revenue_to_cac"] = (
    cac_df["revenue_per_customer"] /
    cac_df["CAC"]
).round(2)

cac_df[[
    "channel",
    "CAC",
    "revenue_per_customer",
    "revenue_to_cac"
]].sort_values(
    by="revenue_to_cac",
    ascending=False
)
```

```python
# Create a Channel-wise Customer Revenue Summary
channel_query = """
SELECT
    c.acquisition_channel,
    cp.customer_id,
    cp.total_revenue
FROM vw_customer_performance cp
JOIN customers c
    ON cp.customer_id = c.customer_id;
"""

channel_df = pd.read_sql(channel_query, engine)

print(channel_df.head())
```

```python
# Aggregate revenue by acquisition channel
channel_summary = (
    channel_df.groupby("acquisition_channel")
    .agg(
        total_customers=("customer_id", "nunique"),
        total_revenue=("total_revenue", "sum"),
        average_revenue=("total_revenue", "mean")
    )
    .reset_index()
)

channel_summary["total_revenue"] = (
    channel_summary["total_revenue"].round(2)
)

channel_summary["average_revenue"] = (
    channel_summary["average_revenue"].round(2)
)

channel_summary = channel_summary.sort_values(
    by="total_revenue",
    ascending=False
)

print(channel_summary)
```

```python
# Calculate Revenue Contribution %
channel_summary["revenue_contribution_pct"] = (
    channel_summary["total_revenue"]
    / channel_summary["total_revenue"].sum()
) * 100

channel_summary["revenue_contribution_pct"] = (
    channel_summary["revenue_contribution_pct"].round(2)
)

print(channel_summary)
```

```python
# Create a Customer Performance Category
import numpy as np

customer_df["customer_category"] = np.where(
    customer_df["total_revenue"] > 50000,
    "High Value",
    np.where(
        customer_df["total_revenue"] >= 20000,
        "Medium Value",
        "Low Value"
    )
)

print(
    customer_df["customer_category"].value_counts()
)
```

```python
# Summarize Customer Categories
category_summary = (
    customer_df.groupby("customer_category")
    .agg(
        customer_count=("customer_id", "nunique"),
        total_revenue=("total_revenue", "sum"),
        average_revenue=("total_revenue", "mean")
    )
    .reset_index()
)

category_summary = category_summary.round(2)

print(category_summary)
```
